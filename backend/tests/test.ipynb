```python
import os
import sys
import json
import time

# Ensure backend path is configured
backend_path = os.path.dirname(os.path.abspath("."))
if backend_path not in sys.path:
    sys.path.insert(0, backend_path)

# Clear import cache for updates
for mod in list(sys.modules.keys()):
    if any(p in mod for p in ('core', 'schemas', 'extraction')):
        sys.modules.pop(mod, None)

from core import settings
from extraction import route_and_extract

papers_dir = os.path.join(backend_path, "papers", "research_papers")
# Target folder changed to complete_phase_1_test
output_dir = os.path.join(settings.storage_dir, "complete_phase_1_test")
os.makedirs(output_dir, exist_ok=True)

if not os.path.exists(papers_dir):
    print(f"Error: Papers directory not found: {papers_dir}")
else:
    pdf_files = [f for f in os.listdir(papers_dir) if f.lower().endswith(".pdf")]
    print(f"Found {len(pdf_files)} PDF documents in corpus.")
    print("🚀 Starting complete Phase-1 Batch Extraction Router...\n")
    
    start_time = time.time()
    success_count = 0
    fail_count = 0
    
    for idx, filename in enumerate(sorted(pdf_files)):
        pdf_path = os.path.join(papers_dir, filename)
        base_name = os.path.splitext(filename)[0]
        
        print(f"[{idx+1}/{len(pdf_files)}] Processing {filename}...")
        try:
            # Execute the router (default local GROBID port 8070)
            result = route_and_extract(pdf_path)
            
            # Save the parsed routed result
            out_name = f"{base_name}_routed.json"
            out_path = os.path.join(output_dir, out_name)
            
            with open(out_path, "w", encoding="utf-8") as f:
                json.dump(result, f, indent=4)
                
            if result.get("valid"):
                success_count += 1
                print(f"   ✓ Success | Saved: {out_name} | Parsers: {result['selected_parsers']}")
            else:
                fail_count += 1
                print(f"   ✗ Invalid PDF | Saved: {out_name} | Error: {result.get('error_message')}")
                
        except Exception as e:
            fail_count += 1
            print(f"   💥 Pipeline Crash | Error: {e}")
            
    duration = time.time() - start_time
    print("\n🏁 BATCH RUN COMPLETE:")
    print(f"  - Total processed: {len(pdf_files)} papers")
    print(f"  - Success rate:    {success_count} / {len(pdf_files)}")
    print(f"  - Failures:        {fail_count}")
    print(f"  - Total Duration:  {duration:.2f} seconds")
    print(f"  - Results saved to: {output_dir}")
```

```python
import os
import sys
import json
import time

# Ensure backend path is configured
backend_path = os.path.dirname(os.path.abspath("."))
if backend_path not in sys.path:
    sys.path.insert(0, backend_path)

# Clear import cache to pick up the updated merger code
for mod in list(sys.modules.keys()):
    if any(p in mod for p in ('core', 'schemas', 'extraction')):
        sys.modules.pop(mod, None)

from core import settings
from schemas.canonical_paper import PaperDocument
from extraction import merge_extractions, validate_paper_document

storage_dir = os.path.join(settings.storage_dir, "complete_phase_2_test")
report_path = os.path.join(
    os.path.dirname(settings.backend_dir), 
    "docs", "backend_docs", "Tests_docs", "Phase_3_test.md"
)

if not os.path.exists(storage_dir):
    print(f"Error: Storage directory not found: {storage_dir}")
else:
    subfolders = [f for f in os.listdir(storage_dir) if os.path.isdir(os.path.join(storage_dir, f))]
    print(f"Refusing to run full pipeline. Running ultra-fast merger + validator cache audit on {len(subfolders)} papers...\n")
    
    start_time = time.time()
    validation_records = []
    total_warnings = 0
    total_errors = 0
    total_success = 0
    
    for idx, folder_name in enumerate(sorted(subfolders)):
        folder_path = os.path.join(storage_dir, folder_name)
        routed_file = f"{folder_name}_routed.json"
        routed_path = os.path.join(folder_path, routed_file)
        
        if not os.path.exists(routed_path):
            continue
            
        try:
            # 1. Load cached routed data
            with open(routed_path, "r", encoding="utf-8") as f:
                routed_result = json.load(f)
                
            # 2. Re-merge using the updated fallback rules
            paper_doc = merge_extractions(routed_result)
            
            # Save the updated canonical file
            canonical_path = os.path.join(folder_path, f"{folder_name}_canonical.json")
            with open(canonical_path, "w", encoding="utf-8") as f:
                f.write(paper_doc.model_dump_json(indent=4))
                
            # 3. Run validator checks
            quality_report = validate_paper_document(paper_doc)
            
            # Save validation report
            quality_path = os.path.join(folder_path, f"{folder_name}_quality.json")
            with open(quality_path, "w", encoding="utf-8") as f:
                f.write(quality_report.model_dump_json(indent=4))
                
            warnings = 0
            errors = 0
            for metric in quality_report.scorecard.values():
                if metric.status == "WARNING":
                    warnings += 1
                elif metric.status == "ERROR":
                    errors += 1
                    
            total_warnings += warnings
            total_errors += errors
            if quality_report.valid:
                total_success += 1
                
            validation_records.append({
                "Paper ID": f"`{paper_doc.paper_id}`",
                "Filename": f"{folder_name}.pdf",
                "Status": "✓ Valid" if quality_report.valid else "✗ Blocked",
                "Errors": errors,
                "Warnings": warnings,
                "Quality Score": f"{len(quality_report.scorecard) - errors - warnings}/{len(quality_report.scorecard)}"
            })
            
        except Exception as e:
            print(f"   💥 Crash on {folder_name} | Error: {e}")
            
    duration = time.time() - start_time
    success_rate = (total_success / len(subfolders)) * 100 if subfolders else 0
    
    # 4. Overwrite Scorecard Report
    md = [
        "# Phase 3 Deterministic Extraction Validation — Scorecard Report\n",
        "This report summarizes the deterministic quality validation scorecard for all 29 research papers, "
        "checking titles, section ordering, text duplication, page coverage, tables, and captions.\n",
        "## 📊 Quality Summary Statistics",
        f"- **Total Papers Audited**: {len(subfolders)}",
        f"- **Clean Valid Ingestions**: {total_success} ({success_rate:.1f}%)",
        f"- **Extraction Errors Flagged**: {total_errors}",
        f"- **Linguistic/Layout Warnings**: {total_warnings}",
        f"- **Audit Duration**: {duration:.2f} seconds\n",
        "## 📋 Quality Scorecard Ledger",
        "| Paper ID | Filename | Status | Errors | Warnings | Quality Score (Passed/Total) |",
        "| :--- | :--- | :---: | :---: | :---: | :---: |"
    ]
    for r in validation_records:
        md.append(
            f"| {r['Paper ID']} | {r['Filename']} | {r['Status']} | {r['Errors']} | {r['Warnings']} | {r['Quality Score']} |"
        )
        
    with open(report_path, "w", encoding="utf-8") as f:
        f.write("\n".join(md))
        
    print(f"🎉 Validation Audit Complete in {duration:.2f} seconds!")
    
    # Render HTML
    from IPython.display import HTML
    html = [
        "<h3 style='font-family:sans-serif; margin-bottom:8px;'>📋 Updated Ingestion Quality Audit</h3>",
        "<table style='width:100%; border-collapse:collapse; font-family:monospace; font-size:12px; border:1px solid #ddd;'>"
    ]
    html.append(
        "<tr style='background-color:#1e1e2e; color:#cdd6f4;'>"
        "<th style='border:1px solid #313244; padding:6px; text-align:left;'>Paper ID</th>"
        "<th style='border:1px solid #313244; padding:6px; text-align:left;'>Filename</th>"
        "<th style='border:1px solid #313244; padding:6px; text-align:center;'>Status</th>"
        "<th style='border:1px solid #313244; padding:6px; text-align:center;'>Errors</th>"
        "<th style='border:1px solid #313244; padding:6px; text-align:center;'>Warnings</th>"
        "<th style='border:1px solid #313244; padding:6px; text-align:center;'>Score</th>"
        "</tr>"
    )
    for r in validation_records:
        bg = "#182a1e" if r["Errors"] == 0 else "#3b1a1a"
        fg = "#a6e3a1" if r["Errors"] == 0 else "#f38ba8"
        html.append(
            f"<tr style='background-color:{bg}; color:{fg};'>"
            f"<td style='border:1px solid #313244; padding:5px;'>{r['Paper ID']}</td>"
            f"<td style='border:1px solid #313244; padding:5px;'>{r['Filename']}</td>"
            f"<td style='border:1px solid #313244; padding:5px; text-align:center;'>{r['Status']}</td>"
            f"<td style='border:1px solid #313244; padding:5px; text-align:center;'>{r['Errors']}</td>"
            f"<td style='border:1px solid #313244; padding:5px; text-align:center;'>{r['Warnings']}</td>"
            f"<td style='border:1px solid #313244; padding:5px; text-align:center;'>{r['Quality Score']}</td>"
            f"</tr>"
        )
    html.append("</table>")
    display(HTML("".join(html)))
```
